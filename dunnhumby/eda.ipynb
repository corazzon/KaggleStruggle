## [Dunnhumby - The Complete Journey](https://www.kaggle.com/datasets/frtgnn/dunnhumby-the-complete-journey)

### 개요
- **Dunnhumby - The Complete Journey** 데이터셋은 미국 소매 유통사의 실제 고객 구매 데이터를 기반으로 한 대규모 거래 데이터셋입니다.
- 약 **2,500 가구**의 2년간의 구매 내역이 포함되어 있으며, 각 가구는 해당 소매점의 단골 고객입니다.
- 모든 가구의 전체 구매 내역(카테고리 제한 없음)이 포함되어 있으며, 일부 가구에 대해서는 인구통계 정보와 직접 마케팅(캠페인, 쿠폰) 이력도 제공됩니다.

### 주요 데이터 구성
데이터셋은 여러 개의 테이블(파일)로 구성되어 있습니다.

| 테이블명              | 설명                                                         |
|----------------------|-------------------------------------------------------------|
| transaction_data     | 각 가구의 상세 구매 내역 (구매일, 상품, 수량, 금액 등)        |
| hhdemographic        | 가구의 인구통계 정보 (연령, 소득, 가족 구성 등)               |
| campaign_table       | 각 가구가 받은 마케팅 캠페인 정보                            |
| campaign_desc        | 캠페인별 기간 및 상세 설명                                   |
| coupon               | 캠페인에 포함된 쿠폰 정보 및 해당 쿠폰이 적용되는 상품 정보   |
| coupon_redempt       | 각 가구가 실제로 사용한 쿠폰 내역                            |
| product              | 판매 상품의 상세 정보 (브랜드, 카테고리, 라벨 등)            |
| causal_data          | 특정 상품이 주간 전단지나 매장 내 프로모션에 노출된 정보      |

### 활용 예시
- **고객 세분화 및 행동 분석**: 고객별 구매 패턴, 재방문 빈도, 지출 변화 등 분석
- **마케팅 효과 분석**: 캠페인 및 쿠폰 발송이 실제 구매에 미치는 영향 측정
- **프로모션 효과 분석**: 특정 상품의 전단지/매장 프로모션이 판매에 미치는 영향 평가
- **인구통계 기반 분석**: 연령, 소득, 가족 구성 등 인구통계 특성이 구매 행동에 미치는 영향 연구

### 데이터셋 특징 및 활용 난이도
- 데이터 테이블이 다수이고, 관계형 구조로 이루어져 있어 **중급 이상 데이터 분석/머신러닝** 실습 및 연구에 적합합니다.
- 실제 소매 유통 환경에서의 고객 행동, 마케팅, 프로모션 효과 등을 정밀하게 분석할 수 있습니다[2][3].

### 활용 가능한 분석 질문 예시
- 시간이 지남에 따라 지출이 증가/감소하는 고객은 누구이며, 이들의 특징은 무엇인가?
- 어떤 상품 카테고리에서 지출이 빠르게 늘어나거나 줄어드는가?
- 특정 인구통계 요인이 구매 행동이나 특정 카테고리와의 연관성이 있는가?
- 직접 마케팅(캠페인, 쿠폰)이 고객 참여와 구매에 미치는 영향은 무엇인가?

```python
import os
import pandas as pd

data_dir = 'data'
csv_files = [f for f in os.listdir(data_dir) if f.endswith('.csv')]

for csv_file in csv_files:
    var_name = os.path.splitext(csv_file)[0]
    df = pd.read_csv(os.path.join(data_dir, csv_file))
    globals()[var_name] = df
    print(f"=== {var_name} ===")
    df.info()
    print("\n")
```
